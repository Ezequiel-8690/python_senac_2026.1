# Uso de import

```python
import random
segredo=random.randint(0,10)
tentativas=0
while tentativas < 3:
    numero=int(input('Digite um número:'))
    if numero == segredo:
        print('você acertou')
        break
    else:
        print('Você errou, tente novamente')
    tentativas=tentativas+1
print(f'O número segredo era:{segredo}.')
```
