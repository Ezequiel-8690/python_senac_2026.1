8️⃣: Multiplicar cada elemento do vetor por 2 e armazenar em um novo vetor

📜 Explicação:
Passo 1: Criamos o vetor original.
Passo 2: Definimos o valor pelo qual cada elemento será multiplicado.
Passo 3: Usamos list comprehension para criar uma nova lista, multiplicando cada elemento.

```python
lista_original=[1,3,5,9,10,12,50,40,60]
mult= 2
lista_mult=[elemento * mult for elemento in lista_original]
print(f'Vetor multipilcado:{lista_mult}',end=' ')
```

```python
for elemento in lista_mult:
    lista_mult.append(elemento*mult)
print(lista_mult)
```

3️⃣ Verificação de números primos

📚 Exercício:
Crie uma tupla com 10 números inteiros e exiba apenas os números primos.

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primos = tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")
```

Crie um programa que recebe o nome e o preço de 5 produtos, armazene-os em uma tupla e exiba uma listagem formatada.

```python
Produtos=(('Arroz',5.99, 5),
          ('Batata',9.99, 9),
          ('Feijão',4.99, 10),
          ('Macarrão',4.50, 9),
          ('Oléo',9.50, 20),
 )
valor_itens=0
valor_total=0

print("LISTA DE PRODUTOS")
for nome, preco, quantidade in Produtos:
    valor_itens= preco * quantidade
    valor_total+=valor_itens
    print(f"{nome:.<20} R$ {preco:.2f} .. quantidade é:{quantidade:02d}!..R${valor_itens:8.2f}")
print(f'O valor total da compra é:{valor_total:5.2f}')
```

# Verificação de números primos

Exercício:
Crie uma tupla com 10 números inteiros e exiba apenas os números primos.

```python
numeros = (2, 7, 10, 13, 17, 22, 29, 31, 40, 53)

def primo(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
primo_lista =[]
for num in numeros:
    if primo(num):
        primo_lista.append(num)

primos = tuple(primo_lista)
print(f'Os números primos são:{primos}')
print(primo_lista)
```
