# Exercício 4: Função fatorial

```python
def fatorial(n):
    if n < 0:
        return'Número inválido para o fatorial.'
    elif n == 0:
        return 1
    else:
        resultado = 1 
        for i in range(1,n +1):
            resultado *= i 
        return resultado
# Exemplo de chamada
print('Fatorial de 5:',fatorial(5))
```

```python
def contar_vogais(texto):
    vogais ='A E I O U a e i o u'
    contador = 0 
    for char in texto:
        if char in vogais:
            contador += 1 
    return contador 
print("Número de vogais em 'Python':", contar_vogais('Python'))
```

# Exercício 10: Função ordenar_por_segundo

```python
def ordenar_por_segundo(lista_tuplas):
    return sorted(lista_tuplas, key=lambda x: x[0])
# Exemplo de chamada
lista_exemplo = [("a", 3), ("b", 1), ("c", 2)]
print("Lista ordenada:", ordenar_por_segundo(lista_exemplo))
```

5️⃣ Escreva um programa Python que use a função main() para calcular o fatorial de um número fornecido pelo usuário.

```python
def fatorial(n):
   if n == 0 or n == 1:
       return 1
   else:
       return n * fatorial(n - 1)

def main():
   num = int(input("Digite um número: "))
   print(f"O fatorial de {num} é {fatorial(num)}")

if __name__ == "__main__":
   main()
```

6️⃣ Desenvolva um programa Python que use a função main() para verificar se um número é primo ou não.

```python
def eh_primo(n):
   if n <= 1:
       return False
   for i in range(2, int(n**0.5) + 1):
       if n % i == 0:
           return False
   return True

def main():
   num = int(input("Digite um número: "))
   if eh_primo(num):
       print(f"{num} é primo.")
   else:
       print(f"{num} não é primo.")

if __name__ == "__main__":
   main()
```

🔠 Desenvolva um programa Python que use a função main() para gerar a sequência de Fibonacci até um número fornecido pelo usuário.

```python
def fibonacci(n):
   sequencia = []
   a, b = 0, 1
   while a <= n:
       sequencia.append(a)
       a, b = b, a + b
   return sequencia

def main():
   num = int(input("Digite um número: "))
   print(f"Sequência de Fibonacci até {num}: {fibonacci(num)}")

if __name__ == "__main__":
   main()
```
