## Uso de dicionarios 

```python
# Criando um dicionario 
telefone={'Alice': '1234-5678',
          'bob': '2245-5698',
          'marcos': '5555-5555',
          'matheus': '2222-2222',
          'laura': '4444-4444',
          'jorge': '8888-8888',}
print(telefone)
```

```python
chaves = ["nome","dt_nascimento","bairro","estado","cep","cidade","sexo"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)

print(usuario)
```

```python
usuario["nome"] = input('Digite seu nome:')
usuario["dt_nascimento"] = input('Digite seu ano de nascimento:')
usuario["bairro"] = input('Digite seu bairro:')
usuario["estado"] = input('Digite seu estado:')
usuario["cep"] = input('Digite seu cep:')
usuario["cidade"] = input('Digite sua cidade:')
usuario["sexo"] = input('Digite seu sexo:')
print(usuario)
```

```python

for chaves,valor in usuario.items():
    print(f'{chaves.capitalize():-<20}:{valor.capitalize()}')
```

```python
matriz = []

for chave, valor in usuario.items():
    # Cria uma linha [chave, valor] com a formatação desejada e adiciona à matriz
    linha = [chave.capitalize(), str(valor).capitalize()]
    matriz.append(linha)

# Exibindo a matriz resultante
print(matriz)
```
