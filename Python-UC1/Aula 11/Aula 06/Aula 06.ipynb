## Estruturas de repetição

### Ultilização do for 

```python
frutas = ['maçâ','banana','cereja']
for fruta in frutas:
    print(fruta)
```

```python
for indice in range(5):
    print(f'indice: {indice} - Posição: {indice+1}')
```

# Ultilização do while 

```python
contador = 0
while (contador <=10):
    print(contador)
    contador += 1
```

```python
matriz =[[1,2,3],
         [4,5,6],
         ['A','B','C'],
         ]
for linha in matriz:
    print("[", end="\t\t")
    for elemento in linha:
        print(elemento, end="\t\t")
    print("]", end=" ")
    print()
```

```python
for i in range(10):
    if i == 5:
        break
    if i % 2 == 0:
        continue
    print(i)
```

# Uso do else no for

```python
for i in range(3):
    print(i)
else:
    print("loop concluído")
```

# Imprimir Números pares e impares

3️⃣ Elabore um código para imprimir todos os números pares e ímpares de 1 a 20 usando for

```python
#Entrada
pares =[]
impares =[]
for numero in range(1,21):
    if (numero % 2 ==0):
        pares.append(numero)
    else:
        impares.append(numero)
print(f'Número pares :{pares}')
print(f'Número impares :{impares}')      
```

#  Imprimir a soma dos números

6️⃣ Elabora um código para imprimir a soma dos números de 1 a 100 usando for

```python
#Entrada
soma = 0
for numero in range(1,101):
    soma =soma+numero
    print(f'O valor da soma é:{soma}')
```

## Soma usando while

```python
soma = 0 
numero = 1 
while numero <= 100:
    soma = soma + numero
    numero = numero + 1
    #print(numero)
    print(soma)
```

# Calcular o fatorial de um número usando while

7️⃣ O programa deve obter um número e em seguida imprimir o valor do fatorial


```python
numero = int(input('Digite um número:'))
fatorial = 1 
numero_aux= numero
while numero_aux >= 1:
    fatorial = fatorial * numero_aux
    numero_aux = numero_aux - 1
print(f'O fatorial de:{numero} e {fatorial}:')

```
