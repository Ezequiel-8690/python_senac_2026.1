# positivo negativo 

1️⃣ Peça ao usuário um número e diga se ele é positivo, negativo ou zero.

```python
#entrada
n1= int(input('Digite um número:'))
if n1 ==0:
    print('Número neutro')
elif n1 >= 1:
    print('Número positivo')
elif n1 <= -1:
    print('Número negativo')
else:
    print('Opção inválida')
```

# Par ou impar 

4️⃣ Faça um programa que peça um número e diga se ele é par ou ímpar.

```python
p= int(input('Digite um número:'))
if p == 0:
    print('Seu número é ZERO')
elif p % 2 ==0:
    print(f'Seu número é:{p} PAR')
else:
    print(f'Seu número é:{p} IMPAR')
```

# Calculadora IMC 

7️⃣ Desenvolva um programa que calcule o IMC de uma pessoa e exiba sua classificação.

```python
peso= float(input('Digite seu peso:'))
altura = float(input('Digite sua altura:'))
imc = peso /(altura**2)
print(f'Seu imc é:{imc:.2f}')
if imc <= 18.5:
    print('Abaixo do peso')
elif imc == 18.5 and 24.9:
    print('Peso ideial')
elif imc >= 25 and 29.9:
    print('Acima do peso')
elif imc <= 30:
    print('Obeso')
```

# Acesso 

5️⃣ Crie um programa que peça um nome de usuário e uma senha. Se o nome for "admin" e a senha "1234", mostre "Acesso permitido". Caso contrário, "Acesso negado".

```python
#Entrada
login ='admin'
senha = 1234
#Processo
p = str(input('Digite o usuário:'))
p2 = int(input('Digite sua senha:'))
#condições
if p == login:
    print('Nome correto ')
elif p2 == senha:
    print('Senha correto')
else:
    print('Acesso negado')    
```
