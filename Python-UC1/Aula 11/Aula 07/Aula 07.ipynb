# Funções nomeadas  

```python
def potencia(numero, expoente):
    resultado = numero ** expoente
    return resultado 
```

```python
numero = 2 
expoente = 4 
print(potencia(numero,expoente))
```

```python
n1 = 2
exp = 8
print(potencia(expoente=exp,numero=n1))
```
