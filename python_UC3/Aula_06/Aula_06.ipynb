### 1. Armazenamento Temporário (Em Memória)


Essas estruturas residem na memória RAM enquanto o programa está rodando. Os dados são perdidos assim que a aplicação é encerrada.

Listas (list): Sequência ordenada e mutável de elementos. Permite itens duplicados.

Métodos principais: append() (adiciona ao fim), pop() (remove pelo índice ou o último), remove() (remove a primeira ocorrência de um valor).

Situação de uso: Filas de tarefas, históricos de navegação temporários, listas de itens em um carrinho de compras em andamento.

Tuplas (tuple): Sequência ordenada, porém imutável (não pode ser alterada após criada).

Métodos principais: count() (conta ocorrências), index() (retorna o índice de um valor). Métodos de alteração (append, remove) não existem.

Situação de uso: Coordenadas geográficas fixas, configurações que não devem mudar durante a execução, chaves compostas em dicionários.

Dicionários (dict): Coleção de pares chave-valor (key: value). Chaves são únicas e imutáveis.

Métodos principais: update() (adiciona/atualiza chaves), pop() (remove uma chave e retorna seu valor), get() (busca segura por uma chave).

Situação de uso: Mapear perfis de usuários por ID, armazenar propriedades de um objeto (como nome, idade, preço), caches rápidos em memória.

Conjuntos (set): Coleção não ordenada de elementos únicos (sem duplicatas).

Métodos principais: add() (adiciona elemento), remove() / discard() (remove elemento), além de operações matemáticas como union() e intersection().

Situação de uso: Remover duplicatas de uma lista, verificar tags únicas, verificar permissões de acesso rapidamente.


### Métodos em Listas (list) — Mutáveis e Ordenadas
Como as tuplas não podem ser alteradas após a criação, elas não possuem métodos de modificação (como append, remove ou pop). Seus métodos servem apenas para consulta.

```python
###
# Criando uma lista de produtos
produtos = ["Brigadeiro", "Beijinho", "Pudim"]

# 1. append(item): Adiciona um item ao final da lista
produtos.append("Coxinha de Ninho")
print("Após append:", produtos) 
# Saída: ['Brigadeiro', 'Beijinho', 'Pudim', 'Coxinha de Ninho']

# 2. insert(indice, item): Insere um item em uma posição específica
produtos.insert(1, "Casadinho") # Entra na posição 1
print("Após insert:", produtos) 
# Saída: ['Brigadeiro', 'Casadinho', 'Beijinho', 'Pudim', 'Coxinha de Ninho']

# 3. pop(indice): Remove e retorna o item de um índice (padrão é o último)
removido = produtos.pop() # Remove "Coxinha de Ninho"
print(f"Item removido com pop: {removido}")

# 4. remove(valor): Remove a primeira ocorrência do valor especificado
produtos.remove("Beijinho")
print("Após remove:", produtos) 
# Saída: ['Brigadeiro', 'Casadinho', 'Pudim']

# 5. sort(): Ordena os itens em ordem alfabética ou numérica
produtos.sort()
print("Após sort:", produtos) 
# Saída: ['Brigadeiro', 'Casadinho', 'Pudim']
```

### Métodos em Tuplas (tuple) — Imutáveis
Como as tuplas não podem ser alteradas após a criação, elas não possuem métodos de modificação (como append, remove ou pop). Seus métodos servem apenas para consulta.

```python
# Coordenadas geográficas ou configurações fixas
coordenada = (-22.505, -43.179, -22.505) # Permite valores repetidos

# 1. count(valor): Conta quantas vezes um valor aparece na tupla
quantidade = coordenada.count(-22.505)
print(f"O valor -22.505 aparece {quantidade} vezes.") 
# Saída: 2

# 2. index(valor): Retorna o índice da primeira ocorrência do valor
posicao = coordenada.index(-43.179)
print(f"O valor -43.179 está no índice: {posicao}") 
# Saída: 1
```

### Métodos em Dicionários (dict) — Chave e Valor
Os dicionários armazenam dados associados por chaves únicas.

```python
# Dados de um perfil de usuário
usuario = {"nome": "Carlos", "idade": 28, "cidade": "Petrópolis"}

# 1. get(chave): Retorna o valor de forma segura (se não existir, retorna None em vez de erro)
print("Nome:", usuario.get("nome"))
print("Telefone:", usuario.get("telefone", "Não cadastrado")) # Retorna valor padrão se não achar

# 2. update(outro_dicionario): Adiciona ou atualiza múltiplos pares de chave-valor de uma vez
usuario.update({"idade": 29, "profissao": "Desenvolvedor"})
print("Após update:", usuario)

# 3. pop(chave): Remove a chave especificada e retorna o seu valor
cidade_removida = usuario.pop("cidade")
print(f"Chave 'cidade' removida, valor era: {cidade_removida}")
print("Dicionário final:", usuario)
```

### Métodos em Conjuntos (set) — Elementos Únicos e Não Ordenados
Perfeitos para operaçoes matemáticas de conjuntos e eliminação automática de duplicatas.

```python
# Tags de interesse (note que repeti "python")
tags = {"python", "html", "css", "python"}
print("Set sem duplicatas:", tags) 
# Saída: {'python', 'html', 'css'} (a ordem pode variar)

# 1. add(item): Adiciona um novo elemento ao conjunto
tags.add("bootstrap")
print("Após add:", tags)

# 2. remove(item): Remove um elemento (gera erro se o elemento não existir)
tags.remove("html")
print("Após remove:", tags)

# 3. discard(item): Remove um elemento (ignora silenciosamente se ele não existir)
tags.discard("javascript") # Não faz nada, mas não quebra o código

# 4. union(outro_set): Junta dois conjuntos unindo os elementos
tags_extras = {"git", "vscode"}
todas_as_tags = tags.union(tags_extras)
print("União dos sets:", todas_as_tags)
```

### 2. Armazenamento Persistente (Após o Encerramento)

Essas formas gravam os dados em arquivos no disco rígido ou em bancos de dados, garantindo que eles continuem disponíveis na próxima vez que o programa for aberto.

Arquivos de Texto (.txt): Leitura e escrita linha por linha.

Como usar: Utiliza a função nativa open(arquivo, 'w' ou 'r').

Situação de uso: Logs simples do sistema, arquivos de configuração rápidos ou notas brutas.

Arquivos JSON (.json): Formato padrão para intercâmbio de dados estruturados (muito similar aos dicionários e listas do Python).

Como usar: Através da biblioteca nativa import json. Usa json.dump() para salvar e json.load() para ler.

Situação de uso: Salvar configurações de usuário, estado de jogos leves, perfis ou pequenos bancos de dados locais.

Módulo Pickle (.pickle / .dat): Serialização nativa do Python para salvar quase qualquer objeto complexo em formato binário (listas, dicionários, instâncias de classes).

Como usar: import pickle. Usa pickle.dump() e pickle.load().

Situação de uso: Salvar modelos de Machine Learning treinados, estados complexos de objetos sem precisar reescrever parsers.

Banco de Dados Relacional (SQLite): Banco de dados leve, embutido no próprio Python, estruturado em tabelas com linhas e colunas.

Como usar: Através da biblioteca nativa import sqlite3, executando comandos SQL (SQLAlchemy é uma alternativa mais avançada).

Situação de uso: Aplicações que exigem buscas rápidas, múltiplos registros estruturados, relatórios e integridade relacional (ex: sistema de cadastro de uma doceria).

### Métodos em Arquivos JSON (Persistência em Disco)
Para persistir estruturas complexas (como dicionários e listas aninhadas) em arquivos físicos no computador:

```python
import json

carrinho_compras = [
    {"item": "Brigadeiro", "preco": 3.50},
    {"item": "Beijinho", "preco": 3.00}
]

# 1. json.dump(): Escreve (salva) os dados estruturados diretamente em um arquivo no disco
with open("carrinho.json", "w", encoding="utf-8") as arquivo:
    json.dump(carrinho_compras, arquivo, ensure_ascii=False, indent=4)

print("Arquivo JSON salvo com sucesso!")


# 2. json.load(): Lê os dados de volta do arquivo do disco para a memória RAM
with open("carrinho.json", "r", encoding="utf-8") as arquivo:
    dados_recuperados = json.load(arquivo)

print("Dados lidos do arquivo JSON:", dados_recuperados)

# Como o que foi carregado vira uma lista normal do Python, 
# você pode usar métodos comuns de lista, como .append()
dados_recuperados.append({"item": "Pudim", "preco": 8.00})
print("Lista após aplicar .append() nos dados recuperados:", dados_recuperados)
```

# segunda pesquisa sobre formas de dados 

No Python, uma lista é uma coleção de itens armazenados em uma sequência específica.
Para visualizar, imagine uma prateleira com gavetas numeradas. Cada gaveta pode guardar um objeto (um número, um texto, ou até outra prateleira). A contagem das gavetas no Python começa sempre do zero (0, 1, 2, 3...). No código, as listas são representadas por colchetes [].

## Características

Para que você saiba exatamente com o que está lidando, anote as quatro características fundamentais de uma lista:

1- Ordenada: A lista mantém a ordem exata em que você inseriu os itens. O primeiro item que entra é o primeiro da fila.

2- Mutável: Após criar uma lista, você pode adicionar, remover ou alterar os itens dela sem precisar criar uma nova lista do zero.

3- Heterogênea: Uma mesma lista pode armazenar tipos de dados diferentes. Você pode ter um texto, um número e um valor booleano na mesma lista (embora, por boas práticas, devamos agrupar itens do mesmo tipo).

4- Dinâmica: O tamanho da lista não é fixo. Ela cresce e encolhe conforme a necessidade da sua aplicação, gerenciando a memória automaticamente.


### Quando utilizar
A comunicação assertiva exige que você saiba o que usar e o que NÃO usar.

Use listas quando:

A ordem dos elementos for importante (ex: histórico de ações, fila de processamento).
Você precisar adicionar ou remover itens frequentemente.
Você precisar acessar itens por sua posição (índice).
Houver possibilidade de itens repetidos na sua coleção.

NÃO use listas quando:

Você precisar de pares de chave e valor (Use Dicionários / dict).
Você precisar garantir que não haja itens repetidos (Use Conjuntos / set).
Você precisar de imutabilidade para garantir que os dados não sejam alterados acidentalmente (Use Tuplas / tuple).

## Principais Métodos


Cenário inicial: A equipe base do projeto Django.
Os índices são: Ana(0), Bruno(1), Carla(2)
equipe = ['Ana', 'Bruno', 'Carla']

1- append(item) 
O que faz: Adiciona um item ao final da lista. É o método mais rápido para inserções. 

O desenvolvedor Daniel foi contratado e vai entrar no final da equipe.
equipe.append('Daniel')
Resultado atual: ['Ana', 'Bruno', 'Carla', 'Daniel']

2- insert(indeice, item)
O que faz: Adiciona um item em uma posição específica (índice), empurrando os outros para a direita.

O Eduardo é o nosso Arquiteto de Software. Ele precisa estar na posição 1, 
logo após a Ana, para ter prioridade na lista de contatos.
equipe.insert(1, 'Eduardo')
Resultado atual: ['Ana', 'Eduardo', 'Bruno', 'Carla', 'Daniel']

3- remove(item)
O que faz: Remove a primeira ocorrência de um valor específico. Se o valor não existir, o Python gera um erro (por isso, seja assertivo: valide se o item existe antes, ou use blocos try/except).
O Bruno decidiu mudar de área e saiu do projeto. Vamos removê-lo pelo nome.
equipe.remove('Bruno')
Resultado atual: ['Ana', 'Eduardo', 'Carla', 'Daniel']

4- pop(indice)
O que faz: Remove e retorna o item de uma posição específica. Se você não passar nenhum índice, ele remove o último item da lista.

O contrato do Daniel (o último da lista) acabou. 
Vamos removê-lo da equipe e guardar o nome dele em uma variável para o RH.
ex_dev = equipe.pop() 
Resultado atual da equipe: ['Ana', 'Eduardo', 'Carla']
Variável ex_dev agora guarda: 'Daniel'

5- sort()
O que faz: Ordena a lista. Por padrão, ordena em ordem alfabética (para textos) ou crescente (para números). Modifica a lista original.
O departamento de RH exigiu que a lista de funcionários esteja em ordem alfabética.
equipe.sort()
Resultado atual: ['Ana', 'Carla', 'Eduardo']

6- index(item)
O que faz: Retorna o índice (a posição) da primeira ocorrência de um item. Essencial para quando você sabe quem está na lista, mas não sabe onde.

Precisamos ligar para a Carla. Em qual posição (índice) ela está agora?
posicao_carla = equipe.index('Carla')
Resultado: 1 (Pois após a ordenação, Ana é 0, Carla é 1 e Eduardo é 2)

7- clear()
O que faz: Remove todos os itens da lista, deixando-a vazia. A lista continua existindo, mas sem conteúdo.
O projeto Django foi finalizado com sucesso e entregue. A equipe foi desfeita.
equipe.clear()
Resultado atual: []
